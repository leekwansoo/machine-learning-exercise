✅ STEP 1. 아래 코드 복사

```python
import pandas as pd

file = '어린이코호트 임상데이터.xlsx'

df = pd.read_excel(file, sheet_name=0, engine='openpyxl')

# 기본 통계
desc = df.describe()

# 새 파일 저장
with pd.ExcelWriter('어린이코호트_자동분석리포트.xlsx', engine='openpyxl') as writer:
    df.to_excel(writer, sheet_name='RawData', index=False)
    desc.to_excel(writer, sheet_name='SummaryStats')
```

🔥 방법 2: 현재 Excel 파일에 직접 추가
👉 기존 파일을 유지하면서 시트만 추가

```python
with pd.ExcelWriter(
    '어린이코호트 임상데이터.xlsx',
    engine='openpyxl',
    mode='a',
    if_sheet_exists='replace'
) as writer:
    
    desc.to_excel(writer, sheet_name='SummaryStats')
```
