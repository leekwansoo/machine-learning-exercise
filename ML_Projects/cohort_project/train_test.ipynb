## Model Quality Validation and Leakage Checks
These checks are added to validate whether perfect scores are genuine or caused by leakage/train-test overlap.

```python
# Evaluate classification performance correctly and check for leakage/overlap
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score, roc_auc_score,
    confusion_matrix, classification_report
)
from sklearn.model_selection import StratifiedKFold, cross_val_score

def evaluate_classifier(X_train, X_test, y_train, y_test, dataset_name):
    clf = RandomForestClassifier(random_state=42)
    clf.fit(X_train, y_train)
    y_pred = clf.predict(X_test)

    print(f"\n=== {dataset_name} Classification Metrics ===")
    print(f"Accuracy : {accuracy_score(y_test, y_pred):.4f}")
    print(f"Precision: {precision_score(y_test, y_pred, average='weighted', zero_division=0):.4f}")
    print(f"Recall   : {recall_score(y_test, y_pred, average='weighted', zero_division=0):.4f}")
    print(f"F1 Score : {f1_score(y_test, y_pred, average='weighted', zero_division=0):.4f}")

    if len(np.unique(y_test)) == 2 and hasattr(clf, "predict_proba"):
        y_prob = clf.predict_proba(X_test)[:, 1]
        print(f"ROC-AUC  : {roc_auc_score(y_test, y_prob):.4f}")

    print("Confusion Matrix:")
    print(confusion_matrix(y_test, y_pred))
    print("Classification Report:")
    print(classification_report(y_test, y_pred, zero_division=0))

    cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
    X_all = pd.concat([X_train, X_test], axis=0)
    y_all = pd.concat([y_train, y_test], axis=0)
    cv_f1 = cross_val_score(clf, X_all, y_all, cv=cv, scoring="f1_weighted")
    print(f"5-Fold CV F1 (mean +- std): {cv_f1.mean():.4f} +- {cv_f1.std():.4f}")
    return clf

def duplicate_overlap_count(X_train, X_test):
    train_hash = pd.util.hash_pandas_object(X_train, index=False)
    test_hash = pd.util.hash_pandas_object(X_test, index=False)
    return np.intersect1d(train_hash.values, test_hash.values).size

def target_like_columns(X, target_name="고위험군"):
    t = target_name.lower()
    return [col for col in X.columns if t in str(col).lower()]

# Female dataset checks
female_model_checked = evaluate_classifier(X1_train, X1_test, y1_train, y1_test, "Female")
female_overlap = duplicate_overlap_count(X1_train, X1_test)
female_target_like = target_like_columns(X1)

print("\nFemale leakage checks:")
print("Duplicate train-test rows:", female_overlap)
print("Target-like columns in features:", female_target_like if female_target_like else "None")

# Male dataset checks
male_model_checked = evaluate_classifier(X2_train, X2_test, y2_train, y2_test, "Male")
male_overlap = duplicate_overlap_count(X2_train, X2_test)
male_target_like = target_like_columns(X2)

print("\nMale leakage checks:")
print("Duplicate train-test rows:", male_overlap)
print("Target-like columns in features:", male_target_like if male_target_like else "None")
```

## Visualization

```python
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, roc_auc_score

def plot_roc_curve(true_y, y_pred):
    """
    plots the roc curve based on the predicted probabilities
    """

    fpr, tpr, thresholds = roc_curve(true_y, y_pred)
    plt.plot(fpr, tpr)
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
```

```python
# based on previous model quality validation generate graphics
import matplotlib.pyplot as plt
import joblib
from sklearn.metrics import roc_curve, roc_auc_score

# ROC-AUC curve for female dataset
female_model = joblib.load("female_model.joblib")
y1_score = female_model.predict_proba(X1_test)[:, 1]
plot_roc_curve(y1_test, y1_score)
print(f'model 1 AUC score: {roc_auc_score(y1_test, y1_score)}')
```
